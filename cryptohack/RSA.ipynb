```python
# Ron was Wrong, Whit is Right

from Crypto.Cipher import PKCS1_OAEP
from Crypto.PublicKey import RSA
from Crypto.Util.number import long_to_bytes, inverse

# if we have many different public keys and all of them used to encrypt same message
# we need to factorize one of them, first which come to my mind is checking whether they have common factors

keys = []
exponents = []
ciphertexts = []

for k in range(1, 51): 
    with open(f'RonwasWrongWhitisRight/{k}.pem') as f:
        key = RSA.importKey(f.read())

    with open(f'RonwasWrongWhitisRight/{k}.ciphertext', 'r') as f:
        ciphertext = bytes.fromhex(f.read())

    keys.append(key.n)
    exponents.append(key.e)
    ciphertexts.append(ciphertext)

for x in range(len(keys)):
    for y in range(x+1,len(keys)):
        px = math.gcd(keys[x], keys[y])
        if px != 1:
            qx = keys[x] // px
            if qx != 1:
                print(f"Factors for key {x}: p={px}, q={qx}")
                index = x
                p = px
                q = qx
                break

key = RSA.construct((p*q, exponents[index], inverse(exponents[index], (p-1)*(q-1))))
cipher = PKCS1_OAEP.new(key)
plaintext = cipher.decrypt(ciphertexts[index])
print(plaintext)
```
