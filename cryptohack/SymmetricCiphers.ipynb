```python
# Structure of AES
def bytes2matrix(text):
    """ Converts a 16-byte array into a 4x4 matrix.  """
    return [list(text[i:i+4]) for i in range(0, len(text), 4)]

def matrix2bytes(matrix):
    """ Converts a 4x4 matrix into a 16-byte array.  """
    return bytes(sum(matrix, []))

matrix = [
    [99, 114, 121, 112],
    [116, 111, 123, 105],
    [110, 109, 97, 116],
    [114, 105, 120, 125],
]

print(matrix2bytes(matrix))
```

```python
# Passwords as Keys

from ast import keyword
import os
from pathlib import Path
import requests
from Crypto.Cipher import AES
import hashlib

resp = requests.get("https://gist.githubusercontent.com/wchargin/8927565/raw/d9783627c731268fb2935a731a618aa8e95cf465/words")
resp.raise_for_status()

data = requests.get("http://aes.cryptohack.org/passwords_as_keys/encrypt_flag/").json()
ciphertext = bytes.fromhex(data["ciphertext"])

# server uses hash of random word as key
# just trying every word in the wordlist as key
# '/passwords_as_keys/decrypt' endpoint is not needed here
for word in resp.text.splitlines():
    word = word.strip()
    KEY = hashlib.md5(word.encode()).digest()
    cipher = AES.new(KEY, AES.MODE_ECB)
    decrypted = cipher.decrypt(ciphertext)
    if decrypted.startswith(b'crypto{'):
        print(decrypted.decode())
        break
```
