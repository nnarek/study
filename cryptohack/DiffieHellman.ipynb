```python
from Crypto.Cipher import AES
from Crypto.Util.Padding import pad, unpad
import hashlib


def is_pkcs7_padded(message):
    padding = message[-message[-1]:]
    return all(padding[i] == len(padding) for i in range(0, len(padding)))


def decrypt_flag(shared_secret: int, iv: str, ciphertext: str):
    # Derive AES key from shared secret
    sha1 = hashlib.sha1()
    sha1.update(str(shared_secret).encode('ascii'))
    key = sha1.digest()[:16]
    # Decrypt flag
    ciphertext = bytes.fromhex(ciphertext)
    iv = bytes.fromhex(iv)
    cipher = AES.new(key, AES.MODE_CBC, iv)
    plaintext = cipher.decrypt(ciphertext)

    if is_pkcs7_padded(plaintext):
        return unpad(plaintext, 16).decode('ascii')
    else:
        return plaintext.decode('ascii')
```

```python
# Export-grade

from pwn import * # pip install pwntools
from discretelog import dlog
import json

r = remote("socket.cryptohack.org", 13379)

def json_read(name):
    return json.loads(r.readline().decode().split(f"Intercepted from {name}: ", 1)[1])

# Alice sends supported encryption algorithms
response = json_read("Alice")
print(response)

# checked with lower than 64 bits and Bob does not support them
response["supported"] = ["DH64"] # sending only weakest option

# we need to send supported encryption algorithms to Bob
r.sendline(json.dumps(response).encode()) 
response = json_read("Bob")
assert response["chosen"] == "DH64"
print(response)

# sending chosen algorithm of Bob to Alice
r.sendline(json.dumps(response).encode()) 
response = json_read("Alice")
p = int(response["p"], 16)
g = int(response["g"], 16)
A = int(response["A"], 16)
print(response)

# now I can not send any message on behalf of Bob and Alice
# I only can intercept their messages

# getting B from Bob
response = json_read("Bob")
B = int(response["B"], 16)
print(response)

# Alice calculates shared secret as B^a mod p and encrypts data using it
response = json_read("Alice")
iv = response["iv"]
encrypted_flag = response["encrypted_flag"]
print(response)


# need to solve A = g^a mod p for a

lam = (p-1)//2 # order of multiplicative group mod p is (p-1)/2, which is still big I think and can not brute forced
assert pow(g, lam, p) == 1 

# using discretelog library to solve discrete log, most probably it is using Pollard's rho algorithm
a,_ = dlog(g, A, p)
shared_secret = pow(B, a, p)

print(decrypt_flag(shared_secret, iv, encrypted_flag))
```
