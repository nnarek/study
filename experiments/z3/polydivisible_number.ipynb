```python
from z3 import *

# Find The 10 Digit Number Classic Puzzle And Solution
# https://youtu.be/xX_93cTzM5Y?si=GzQ6LHiwhHksyKp8
N=10 #seems like N=11 have no solution
s = Solver()

digits = [ Int("t_%s" % (j+1)) for j in range(N)  ]
s.add([And(0<=digits[j],digits[j]<10) for j in range(N)])
s.add(Distinct(digits))

numbers = [ Sum([ digits[j+i]*10**j for j in range(N-i)  ]) for i in range(N) ]

s.add([ numbers[j] % (N-j) == 0 for j in range(N)  ])

while True:
    st=s.check()
    if st==sat:
        m=s.model()
        print([m.eval(digits[j]) for j in range(N)])#finds and prove that it is only solution
        s.add(Not(And([m.eval(digits[j])==digits[j] for j in range(N)])))
    else:
        print(st)
        break
```
