```python
from z3 import *

# Can You Solve The "IMPOSSIBLE" Handshake Logic Puzzle?
# https://youtu.be/W_TOq0zs6KM?si=C1yjnmhy6Y_nG7GD

num=10

dim=[num,num]

s=Solver()
is_shaked = Array("is_shaked",IntSort(),IntSort(),BoolSort())



def InsideBoard(x,y,dim):
    return And(0<=x,x<dim[0],0<=y,y<dim[1])

i,j = Ints("i j")
s.add(ForAll([i,j],Implies(InsideBoard(i,j,dim),  And(is_shaked[i,j]==is_shaked[j,i]))))

s.add(ForAll([i],Implies(InsideBoard(i,i,dim),  And(is_shaked[i,i]==False))))

s.add(ForAll([i],Implies(And(InsideBoard(i,i+1,dim),i%2==0),  And(is_shaked[i,i+1]==False))))

s.add(Distinct([Sum([is_shaked[i,j] for j in range(num)]) for i in range(1,num)  ]))

#TODO able to find that solution is 4,4 but able to prove that it is only solution after 160min
while True:
    st=s.check()
    if st==sat:
        m=s.model()
        print([[m.eval(is_shaked[i,j]) for i in range(num)] for j in range(num) ])
        wife_hands_shakers=Sum([is_shaked[0,j] for j in range(num)])
        husbant_hands_shakers=Sum([is_shaked[1,j] for j in range(num)])
        print(m.eval(husbant_hands_shakers))
        print(m.eval(wife_hands_shakers))
        s.add(Not(And(m.eval(husbant_hands_shakers)==husbant_hands_shakers,m.eval(wife_hands_shakers)==wife_hands_shakers)))
    else:
        print(st)
        break





```
