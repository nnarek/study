```python
from z3 import *

# Can you solve this fiendishly difficult math problem?
# https://youtu.be/fLwFuLuh6p0?si=t-mQPMiCLBVEhK9l


s=Solver()
a,b,c,d,e,x,y,area = Reals("a b c d e x y area")
s.add(area==x*y)
s.add([0<var for var in (a,b,c,d,e,x,y,area)])
#s.add(1<a,a<b,b<c,c<d,d<e,e<y,y<x)

s.add(d+e==x)
s.add(a+a+1+e==x)
s.add(a+a+c+c==x)
s.add(b+c+c==x)

s.add(d+a+b==y)
s.add(d+1+c==y)
s.add(e+c==y)

while True:
    st=s.check()
    if st==sat:
        m=s.model()
        print(m.eval(area))
        s.add(m.eval(area)!=area)
    else:
        print(st)
        break

```
