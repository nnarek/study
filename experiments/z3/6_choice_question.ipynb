```python
from z3 import *

# https://youtu.be/h_fLW4xlOyA?si=EwpIlNVhUQyf2Ij1
# Can You Solve This Tricky Multiple Choice Question?


answers = [ Bool("answers_%s" % (j+1)) for j in range(6)  ]
s = Solver()
s.add(answers[0] == And(answers[1:]))
s.add(answers[1] == Not(Or(answers[2:])))
s.add(answers[2] == And(answers[:2]))
s.add(answers[3] == (Sum(answers[:3])==1))
s.add(answers[4] == Not(Or(answers[:4])))
s.add(answers[5] == Not(Or(answers[:5])))


if s.check() == sat:
    m = s.model()
    print([m.eval(answer) for answer in answers])

print(s.check())
```
