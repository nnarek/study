```python
from z3 import *


P = Datatype('P')
P.declare('A')
P.declare('B')
P.declare('C')
P = P.create()

a=8
b=11
c=15
N=(a+b+c)//2
s = Solver()

resting_players = [ Const("resting_player_%s" % (j+1),P) for j in range(N)  ] 

def is_valid_game_set(resting_players):
    conds=[]
    for x in range(len(resting_players)-1):
        next_resting_player=resting_players[x+1]
        current_resting_player=resting_players[x]
        conds.append(current_resting_player!=next_resting_player)

    conds.append(Sum([resting_player!=P.A for resting_player in resting_players])==a)
    conds.append(Sum([resting_player!=P.B for resting_player in resting_players])==b)
    conds.append(Sum([resting_player!=P.C for resting_player in resting_players])==c)
    return conds


s.add(is_valid_game_set(resting_players))

while True:
    st=s.check()
    if st==sat:
        m = s.model()
        print([m.eval(resting_player) for resting_player in resting_players])
        s.add(Not(And([m.eval(resting_player)==resting_player for resting_player in resting_players])))
    else:
        print(st)
        break
```

```python
from z3 import *

# https://www.youtube.com/watch?v=AVUj7O0Uljk&list=PLDZcGqoKA84GZXo_6e4v_gKf7u4UcNgRl&index=36
# Can you solve this hard logic puzzle? Who lost the 4th game?


P = Datatype('P')
P.declare('A')
P.declare('B')
P.declare('C')
P = P.create()

a=8
b=11
c=15
N=(a+b+c)//2
s = Solver()

resting_players = Array("resting_players",IntSort(),P)

def is_valid_game_set(resting_players):
    conds=[]
    for x in range(N-1):
        next_resting_player=resting_players[x+1]
        current_resting_player=resting_players[x]
        conds.append(current_resting_player!=next_resting_player)

    conds.append(Sum([resting_players[x]!=P.A for x in range(N)])==a)
    conds.append(Sum([resting_players[x]!=P.B for x in range(N)])==b)
    conds.append(Sum([resting_players[x]!=P.C for x in range(N)])==c)
    return And(conds)

second_resting_players = Array("second_resting_players",IntSort(),P) 

confident_game=Int("confident_game")

s.add(confident_game<N)
s.add(0<=confident_game)

s.add(is_valid_game_set(resting_players))

s.add(Not(Exists([second_resting_players],And(is_valid_game_set(second_resting_players),resting_players[confident_game]!=second_resting_players[confident_game]))))

while True:
    st=s.check()
    if st==sat:
        m = s.model()
        print(m.eval(confident_game),m.eval(resting_players[confident_game]))
        s.add(m.eval(confident_game)!=confident_game)
    else:
        print(st)
        break
```
