```python
from z3 import *

m = Int('m')
s = Solver()
a,b,c = Ints('a b c')
x,y,z = Ints('x y z')
t1,t2,q,w,e,n = Ints('t1 t2 q w e n')
def cond(x,y,z,m):
    possible_values = [a+b+c==30,a>=2,b>=2,c>=2,Exists([t1,t2],And([Or([t1==a,t1==b,t1==c]),Or([t2==a,t2==b,t2==c]),t1==t2+6]))]
    cnd=ForAll([a,b,c],Implies(And(possible_values),(If(x<=a,x,0)+If(y<=b,y,0)+If(z<=c,z,0))>=m) )
    return And(x>=0,y>=0,z>=0,cnd)

s.add(cond(x,y,z,m))

s.add(Not(Exists([q,w,e,n],And(cond(q,w,e,n),n>m))))
#using s.maximize(m) instead will not work because of some issues inside z3

st=s.check()
if st==sat:
    print(s.model())


```
