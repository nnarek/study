```python
from z3 import *

# Can You Solve The Mind-Twisting Logical List Riddle? 
# https://youtu.be/vgqUISdGbvk?si=i7qQWQdIkoK218ZD
n=10

answers = [ Bool("answers_%s" % (j+1)) for j in range(n)  ]
#stats = [ (n-Sum(answers))==j for j in range(1,n+1)  ]# exaclty j stats are false
stats = [ (n-Sum(answers))>=j for j in range(1,n+1)  ]# at least j stats are false

s = Solver()
s.add([ stats[j]==answers[j] for j in range(n)  ])



while True:
    st=s.check()
    if st==sat:
        m = s.model()
        print([m.eval(answers[j]) for j in range(n)])
        s.add(Not(And([ m.eval(answers[j])==answers[j] for j in range(n)  ])))
    else:
        print(st)
        break
```
