```python
from z3 import *

# Can you solve the 4 foods puzzle?
# https://youtu.be/WMM0b2Sa0T4?si=OYzozy1_b038TN-Z

factor=5
#TODO if factor is 1 then Optimizer works very slowly and can not understand that problem can be simplified by reducing number of peoples and saving their relative quantity same 

s=Optimize()

peoples=[Bools("a_%i b_%i c_%i d_%i" % (i,i,i,i))  for i in range(100//factor)]

s.add(Sum([p[0] for p in peoples])==65//factor)
s.add(Sum([p[1] for p in peoples])==80//factor)
s.add(Sum([p[2] for p in peoples])==80//factor)
s.add(Sum([p[3] for p in peoples])==90//factor)

min=Sum([And(p) for p in peoples])

s.minimize(min)


while True:
    st=s.check()
    if st==sat:
        m=s.model()
        print(m.eval(min*factor))
        break
    else:
        print(st)
        break





```
