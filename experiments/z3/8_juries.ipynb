```python
from z3 import *

#http://qarakusi.am/taskfolder/1063/task/12255

def find_max_participants(num_j,start_num_vars=1):
    current_max=-1

    for num_vars in range(start_num_vars,1000):
        s=Solver()
        vars = [[Bool("vars_%i_%i" % (i,j)) for i in range(num_j)] for j in range(num_vars)]
        for x in range(num_vars):
            for y in range(x+1,num_vars):
                a=vars[x]
                b=vars[y]
                for first_vote in (False,True):
                    for second_vote in (False,True):
                        s.add(Or([And(a[i]==first_vote,b[i]==second_vote) for i in range(num_j)]))


        st=s.check()
        if st==sat:
            current_max=num_vars
            print(num_j,current_max)
        else:
            print(st)
            break
    return current_max

for x in range(9,100):
    print(x,"=",find_max_participants(x,50))


# 4 = 3
# 5 = 4
# 6 = 10
# 7 = 15
# 8 = 35
# 9 = 56
    
# https://oeis.org/A037952
# a(n)=binomial(n, floor((n-1)/2))=a(n-1)*(n-1)/(floor((n-1)/2) + n mod 2)
```
