```python
from z3 import Solver, Int, Or, Distinct, sat, And, If
import z3
from random import randint

z3.set_param('verbose', 1000)
z3.set_param('trace', True)
z3.set_param('stats', True)
z3.set_param('proof', True)

ai = [randint(2**16,2**20) for x in range(10)]
ai=(list)(set(ai))
ao = [ Int("x_%s" % (i+1)) for i in range(len(ai)) ]
sorted_c  = [ ao[i-1] <= ao[i] for i in range(1,len(ai)) ]


equal_c   = [ Or([ ao[i]==ai[j] for i in range(len(ai)) ]) 
             for j in range(len(ai)) ]

sort_c = sorted_c + equal_c

s = Solver()
s.add( sort_c)
if s.check() == sat:
    m = s.model()
    r = [ m.evaluate(ao[i]) for i in range(len(ai))  ]
    print("sorted array: ",r)
else:
    print("failed to solve")



```

```python
from z3 import *

# in the logs I see "merge b c" "merge (g b) (g c)" ... I guess it is merge of eclasses

z3.set_param('verbose', 1000)
z3.set_param('trace', True)
z3.set_param('stats', True)
z3.set_param('proof', True)


A = DeclareSort('A')           
f = Function('f', A, A, A)     
g = Function('g', A, A)     
h = Function('h', A, A)  

a, b, c = Consts('a b c', A)

x, y = Consts('x y', A)
commutativity = ForAll([x, y], f(x, y) == f(y, x))

goal = Implies(b == c, f(h(a), g(c)) == f(g(b), h(a)))

s = Solver()
s.add(commutativity)
s.add(Not(goal))  

result = s.check()
print("Is 'b = c → f (h(a), g(c)) = f (g(b), h(a))' valid?", result == unsat)
```
