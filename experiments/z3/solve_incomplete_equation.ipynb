```python
from z3 import *

# Can You Solve This Logic Puzzle?
# https://youtu.be/e74VYI3Jeo4?si=bbuSYjVTbjuMoWw7



s=Solver()

a,b,c,x = Ints("a b c x")

s.add(3*a+5*b+c==16)
s.add(4*a+7*b+c==30)
s.add(a+b+c==x)

while True:
    st=s.check()
    if st==sat:
        m=s.model()
        print(m.eval(x))
        s.add(Not(And(m.eval(x)==x)))
    else:
        print(st)
        break





```
