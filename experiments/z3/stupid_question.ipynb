```python
from z3 import *

# A test question that stumped many students!
# https://youtu.be/UpS9VdY7Nxg?si=3_fMVRk8p29oWsoW



s=Solver()

b,y,r,p,n = Ints("b y r p n")

s.add(b>0,y>0,r>0,p>0)
s.add(b+y+p+r==n)
s.add(b+14==n)
s.add(y+16==n)
s.add(r+24==n)
s.add(p+12==n)

while True:
    st=s.check()
    if st==sat:
        m=s.model()
        print(m.eval(n))
        s.add(Not(And(m.eval(n)==n)))
    else:
        print(st)
        break





```
