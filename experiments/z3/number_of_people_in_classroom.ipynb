```python
from z3 import *

# Seemingly Impossible Number Of People In The Room Puzzle
# https://youtu.be/mlXmh33yvf0?si=zyMNZHgdPmMm8jFa



solver=Solver()

s,n,a = Ints("s n a")# sum of all ages,number of people,age of Ann

solver.add(1<=n)

solver.add((s+a)*n==s*(n+1)+4*n*(n+1))#(s+a)/(n+1)=s/n+4
solver.add((s+a+a)*n==s*(n+2)+7*n*(n+2))#(s+a+a)/(n+2)=s/n+7

while True:
    st=solver.check()
    if st==sat:
        m=solver.model()
        print(m.eval(n))
        solver.add(Not(And(m.eval(n)==n)))
    else:
        print(st)
        break





```
