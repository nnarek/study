```python
from z3 import *

#there are 5 two sided balls and 5 colors
#same color should be used to color exactly 2 different balls
#how many ways we can color balls?
#order of balls are not important
#in this case it is same as 5!/2

num_of_colors = 5
num_of_balls = 5
B1 = [Int("ball_side1_%i" % i) for i in range(num_of_balls)]
B2 = [Int("ball_side2_%i" % i) for i in range(num_of_balls)]
s = Solver()

s.add(And([B1[i] < B2[i] for i in range(num_of_balls)]))
s.add(And([Sum([b==i for b in B1+B2])==2 for i in range(num_of_colors)]))
s.add(And([B1[i] <= B1[i-1] for i in  range(1,num_of_colors) ] ))

num_of_solutions = 0
while True:
    st=s.check()
    if st==sat:
        m = s.model()
        num_of_solutions += 1
        for i in range(num_of_balls):
            t=[m.eval(B1[i],model_completion=True),m.eval(B2[i],model_completion=True)]
            print(t,end="")
        print()
        s.add(Not(And([m.eval(b,model_completion=True)==b for b in B1+B2 ])))
    else:
        break
print(num_of_solutions)
```
