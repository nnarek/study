```python
from z3 import *

# What Positive Number Doubles When The Last Digit Moves To The First Digit? Riddle For "Geniuses"
# https://youtu.be/1lHDCAIsyb8?si=b8EHOS-GcYgK3vv9

FirstDigit = RecFunction("FirstDigit", IntSort(),IntSort())
n = FreshConst(IntSort())
RecAddDefinition(FirstDigit,[n],If(n<10, n, FirstDigit(n/10)))

InsertFirstDigit = RecFunction("InsertFirstDigit", IntSort(),IntSort(),IntSort())
n = FreshConst(IntSort())
d = FreshConst(IntSort())
RecAddDefinition(InsertFirstDigit,[n,d],If(n<10, d*10+n, InsertFirstDigit(n/10,d)*10+(n%10)))

#TODO quickly finds some solution but finds smallest solution in 22min

s=Solver()

n = Int("n")
s.add(0<n)
s.add(InsertFirstDigit(n,n%10)/10==2*n)

m = Int("m")
s.add(Not(Exists([m],And(m<n,0<m,InsertFirstDigit(m,m%10)/10==2*m))))

while True:
    st=s.check()
    if st==sat:
        m = s.model()
        print(m.eval(n))
        s.add(n!=m.eval(n))
    else:
        print(st)
        break
```
