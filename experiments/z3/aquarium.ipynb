```python
from z3 import *
from pprint import pprint
from fractions import Fraction
from z3lib import *

dim=[6,6]#https://www.puzzle-aquarium.com/?e=MjoxMCw4MjIsMzcy
field_arg=[
    [0 ,1 ,2 ,3 ,4 ,4],
    [0 ,5 ,5 ,6 ,6 ,6],
    [7 ,8 ,8 ,9 ,9 ,10],
    [7 ,7 ,8 ,11,11,10],
    [12,13,13,14,15,16],
    [12,17,17,14,15,16],
]
top=[5,4,4,5,3,2]
right=[4,3,2,5,5,4]
right=right[::-1]


dim=[12,12]#https://play.google.com/store/apps/details?id=com.algotgames.aquarium&hl=en&gl=US
field_arg=[
    [0, 1, 1, 2, 2, 2, 2, 2, 2, 3, 3, 3],
    [0, 0, 4, 4, 2, 5, 6, 7, 3, 3, 3, 8],
    [9, 10,4, 4, 4, 5, 7, 7, 7, 11,12,12],
    [9, 10,10,13,5, 5, 7, 7, 7, 11,11,14],
    [9, 10,10,10,5, 7, 7, 7, 7, 11,15,14],
    [16,16,16,16,17,17,18,19,20,21,21,14],
    [16,16,16,22,23,23,19,19,20,21,21,14],
    [24,24,24,22,23,23,25,25,25,26,21,27],
    [24,22,22,22,22,25,25,25,25,28,27,27],
    [24,29,30,22,22,31,32,33,33,33,33,27],
    [29,29,34,34,22,31,32,35,36,33,37,38],
    [29,29,29,34,34,36,36,36,36,37,37,38]
]
top=  [5,7,9,6,9,6,4,2,3,6,6,7]
right=[5,8,4,2,7,2,7,4,7,7,9,8]
right=right[::-1]


s = Solver()

Field2D = Array2D(IntSort())
Board2D = Array2D(BoolSort())
field  = Const('field', Field2D)
board  = Const('board', Board2D)

#field_arg=[]
field_arg=list(zip(*field_arg[::-1]))
#print(field_arg)
for x in range(len(field_arg)):
    for y in range(len(field_arg[x])):
        s.add(field[x,y]==field_arg[x][y])


def InsideBoard(x,y,field_arg):
    if isinstance(field_arg,list) and isinstance(field_arg[0],list):
        return And(0<=x,x<len(field_arg),0<=y,y<len(field_arg[0]))
    else:
        return And(0<=x,x<field_arg[0],0<=y,y<field_arg[1])

x1, y1,x2,y2=Ints("x1 y1 x2 y2")
s.add(ForAll([x1, y1], Implies(Not(InsideBoard(x1, y1,dim)),   field[x1, y1]==-1)))

s.add(ForAll([x1, y1,x2,y2], Implies(And(InsideBoard(x1,y1,dim),InsideBoard(x2,y2,dim),field[x1, y1]==field[x2, y2]),  
                                And(
                                    Implies(And(y1==y2),                                                board[x1, y1]==board[x2, y2]),
                                    Implies(And(board[x1, y1]==False,board[x2, y2]==True),              y1>y2)))))

for z in range(dim[0]):
    s.add(Sum([board[z,y] for y in range(dim[1])])==top[z])
    s.add(Sum([board[x,z] for x in range(dim[1])])==right[z])
    pass

while True:
    st=s.check()
    if st==sat:
        m = s.model()
        
        solution=[[-1 for x in range(dim[1])] for x in range(dim[0])]
        for x in range(dim[0]):
            for y in range(dim[1]):
                solution[x][y]=m.eval(board[x,y])
        sol=list(zip(*solution))[::-1]
        for row in sol:
            for column in row:
                print(str(column)[0],end=" ")
            print()
        print()
        s.add(Not(And([m.eval(board[x,y])==board[x,y]  for y in range(dim[1]) for x in range(dim[0]) ])))

    else:
        print(st)
        break
```
