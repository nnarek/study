```python
from z3 import *



def find_max_prod(sum):
    current_max=-1

    for num_vars in range(1,sum+1):
        s=Optimize()
        vars = [Int("vars_%i" % i) for i in range(num_vars)]
        s.add(Sum(vars)==sum)
        s.add([0<var for var in vars])
        s.maximize(Product(vars))


        st=s.check()
        if st==sat:
            m=s.model()
            mx=m.eval(Product(vars)).as_long()
            current_max=max(mx,current_max)
        else:
            print(st)
            return -1
    return current_max

for x in range(1,100):
    print(x,find_max_prod(x))#TODO take long time to solve for sum=10 

#can be solved recursivelly if we assume that first number is A then find_max_prod(sum)=A*find_max_prod(sum-A) where A<=4 because starting from 5  find_max_prod(5)>5 and we can write find_max_prod(A)*find_max_prod(sum-A) instead of A*find_max_prod(sum-A) to get higher value
#A=2 or 3 or 4

```
