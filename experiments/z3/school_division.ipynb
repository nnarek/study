```python
from z3 import *

# https://youtu.be/qV4yBJsRkG0?si=Hjz0Uss8HTbhSFxU
# A Long Division Puzzler

#TODO assume that there can be reminder

def create_num(s,name,size):
    z = [Int(name +("_%s" % j)) for j in range(size)]
    for digit in z:
        s.add(And(0<=digit,digit<=9) )
    return z, Sum([z[k]*10**k for k in range(len(z))])

def eq(s,num,val):
    for x in range(len(val)):
        if val[x]!="*":
            s.add(num[len(num)-1-x]==val[x])

s = Solver()

dividend,sdividend = create_num(s,"dividend",6)
divisor,sdivisor = create_num(s,"divisor",3)
quotient,squotient = create_num(s,"quotient",3)

t=[5,4]
subs = [(dividend,sdividend)] + [create_num(s,"sub_%i" % i,t[i]) for i in range(len(t))]
sub = [x[0] for x in subs]
ssub = [x[1] for x in subs]

t=[6,5,4]
mults = [create_num(s,"mult_%i" % i,t[i]) for i in range(len(t))]
mult = [x[0] for x in mults]
smult = [x[1] for x in mults]


s.add(sdividend == sdivisor*squotient)

for x in range(len(quotient)):
    s.add(sdivisor*quotient[x]*10**x == smult[len(smult)-x-1])

for x in range(len(sub)-1):
    s.add(ssub[x]-smult[x]==ssub[x+1])

nums = [divisor,quotient]+sub+mult
vals = ["**9","*53","6*8***","*9***","**4*","***2**","**4**","****"]
snums = [sdivisor,squotient]+ssub+smult

for x in range(len(nums)):
    eq(s,nums[x],vals[x])

while True:
    st=s.check()
    if st==sat:
        m = s.model()
        print([m.eval(num) for num in snums])
        s.add(Not(And([m.eval(num)==num for num in snums])))
    else:
        print(st)
        break
```
