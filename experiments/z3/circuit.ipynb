```python
from z3 import *


s = Solver()

inp = [Bool("inp_%i" % i) for i in range(6)]

def Enc(i):
    return  i[0],Not(i[0])

def Dec(i):
    return  And(Sum(i)==1,Or(i[0],i[2])),And(Sum(i)==1,Or(i[0],i[1]))

not1 = Not(inp[1])
nxor1 = Not(Xor(inp[0],not1))
or1 = Or(inp[2],inp[3])
and1 = And(inp[4],inp[5])
not2 = Not(nxor1)
dec1_out1,dec1_out2 = Enc([and1])
enc1_out1,enc1_out2 = Dec([not2,or1,dec1_out1,dec1_out2])
out = Not(Xor(enc1_out1,enc1_out2))

s.add(out)

for x in range(1000):
    st=s.check()
    if st==sat:
        m = s.model()
        print([m.eval(a,model_completion = True) for a in inp])
        s.add(Not(And([m.eval(a,model_completion = True)==a for a in inp])))
    else:
        print(st)
        break


```
