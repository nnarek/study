```python
from z3 import *

# https://www.youtube.com/watch?v=bBSS2l1zdvA&list=PLDZcGqoKA84GZXo_6e4v_gKf7u4UcNgRl&index=13
# Hardest Multiple Choice Question Ever!

# redefining question 
# how many answers of this question are right
# a)1
# b)2
# c)0
# d)1

right_answer = Int("r")

a = (right_answer==1)
b = (right_answer==2)
c = (right_answer==0)
d = (right_answer==1)

question = (right_answer == Sum(a,b,c,d))

s = Solver()
s.add([question])
if s.check() == sat:
    m = s.model()
    print(m.eval(right_answer))

print(s.check())
```
