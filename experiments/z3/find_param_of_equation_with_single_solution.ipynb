```python
from z3 import *
#from cvc5.pythonic import *

# 1000 Euro Question!
# https://www.youtube.com/watch?v=DEG3XVcmakA

a,x,t = Reals("a x t")

s=Solver()

def equation(a,x):
    return a*x*x-5*x+2==0


s.add(Exists([x],And(equation(a,x),Not(Exists([t],And(equation(a,t),t!=x))))))

#s.add(a==RealVal(25)/8) 
#s.add(a==RealVal(0))

#s.add(a!=RealVal(25)/8)# able to prove that 25/8 are only solutions but can not find 25/8 solution without hint
#s.add(a!=RealVal(0))
while True:
    st=s.check()
    if st==sat:
        m = s.model()
        print(m.eval(a))
        s.add(a != m.eval(a))
    else:
        print(st)
        break
```
