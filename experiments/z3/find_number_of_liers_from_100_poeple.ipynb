```python
from z3 import *

# "Impossible" Logic Puzzle - How Many Liars Are At The Party? 
# https://youtu.be/jqX9nnRiD9g?si=lOkBS14GwvERDYYH
N=100
s = Solver()

is_truth_teller = [ Bool("people_%s" % (j+1)) for j in range(N)  ] 

t1,t2 = Ints("t1 t2")

f = Function("f", IntSort(),IntSort(),BoolSort())
s.add(ForAll([t1,t2],f(t1,t2)==f(t2,t1)))

number_of_shakes = Function("number_of_shakes", IntSort(),IntSort(),IntSort())#(person_id,iter=N,number_of_shakes)
s.add(ForAll([t1],number_of_shakes(t1,-1)==0))
s.add(ForAll([t1,t2],Implies(t2>=0,number_of_shakes(t1,t2)==(number_of_shakes(t1,t2-1)+f(t1,t2)))))

s.add([is_truth_teller[j]==(number_of_shakes(j,N)==j) for j in range(N)])

number_of_liers = Int("number_of_liers")
s.add(number_of_liers == (N-Sum(is_truth_teller)))

while True:
    st=s.check()
    if st==sat:
        m = s.model()
        print([m.eval(is_truth_teller[j]) for j in range(N)])#TODO finds that solution is 99 but can not prove that it is only possible solution at least after 40min
        print(m.eval(number_of_liers))
        s.add(number_of_liers != m.eval(number_of_liers))
    else:
        print(st)
        break
```
